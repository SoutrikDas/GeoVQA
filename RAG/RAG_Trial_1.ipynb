```python
import cohere
import numpy as np

COHERE_API_KEY = # Use your key here
co = cohere.Client(COHERE_API_KEY)

question = "What is best dish in Paris"

try:
    chat_response = co.chat(
        model="command-a-03-2025",
        # single string message:
        message=question,
        # optional: a preamble acts like a system instruction on many SDKs
        preamble="You are a helpful and factual assistant.",
        max_tokens=300
    )
    answer = getattr(chat_response, "text", None) or str(chat_response)
    print("\nCohere answer:\n", answer)

except Exception as e:
    print("Chat request failed:", type(e).__name__, e)
    answer = None

# embed if we have an answer
if answer:
    embed_response = co.embed(
        texts=[answer],
        model="embed-english-v3.0",
        input_type="search_document"
    )
    embeds = np.array(embed_response.embeddings)
    print("\nEmbedding shape:", embeds.shape)
```
